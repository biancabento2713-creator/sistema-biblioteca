```python

import json
import os

# Nome do arquivo que armazenará os livros
ARQUIVO = "biblioteca.json"

# Lista global de livros
livros = []


# Carrega os livros salvos no arquivo JSON
def carregar_livros():
    global livros

    if os.path.exists(ARQUIVO):
        try:
            with open(ARQUIVO, "r", encoding="utf-8") as arquivo:
                livros = json.load(arquivo)
        except (json.JSONDecodeError, OSError):
            livros = []
    else:
        livros = []


# Salva os livros no arquivo JSON
def salvar_livros():
    with open(ARQUIVO, "w", encoding="utf-8") as arquivo:
        json.dump(livros, arquivo, ensure_ascii=False, indent=4)


# Cadastra um novo livro
def cadastrar_livro():
    print("\n--- CADASTRO DE LIVRO ---")

    titulo = input("Título: ").strip()
    autor = input("Autor: ").strip()

    if not titulo or not autor:
        print("Erro: título e autor são obrigatórios!")
        return

    try:
        ano = int(input("Ano de publicação: "))
        if ano < 0 or ano > 9999:
            print("Erro: informe um ano válido!")
            return
    except ValueError:
        print("Erro: digite um ano usando números!")
        return

    livro = {
        "titulo": titulo,
        "autor": autor,
        "ano": ano,
        "disponível": "Sim"
    }

    livros.append(livro)
    salvar_livros()

    print("Livro cadastrado com sucesso!")


# Exibe todos os livros
def listar_livros():
    print("\n--- ACERVO DA BIBLIOTECA ---")

    if not livros:
        print("Nenhum livro cadastrado.")
        return

    for i, livro in enumerate(livros, start=1):
        print(f"\nLivro {i}")
        print(f"Título: {livro['titulo']}")
        print(f"Autor: {livro['autor']}")
        print(f"Ano: {livro['ano']}")
        print(f"Disponível: {livro['disponível']}")


# Busca livros por título ou autor
def buscar_livro():
    print("\n--- BUSCAR LIVRO ---")
    termo = input("Digite o título ou autor: ").strip().lower()

    encontrados = [
        livro for livro in livros
        if termo and (
            termo in livro["titulo"].lower()
            or termo in livro["autor"].lower()
        )
    ]

    if encontrados:
        for livro in encontrados:
            print(f"\nTítulo: {livro['titulo']}")
            print(f"Autor: {livro['autor']}")
            print(f"Ano: {livro['ano']}")
            print(f"Disponível: {livro['disponível']}")
    else:
        print("Nenhum livro encontrado.")


# Realiza o empréstimo de um livro
def emprestar_livro():
    print("\n--- EMPRÉSTIMO ---")
    titulo = input("Digite o título do livro: ").strip().lower()

    for livro in livros:
        if livro["titulo"].lower() == titulo:
            if livro["disponível"] == "Sim":
                livro["disponível"] = "Não"
                salvar_livros()
                print("Empréstimo realizado com sucesso!")
            else:
                print("Este livro já está emprestado.")
            return

    print("Livro não encontrado.")


# Registra a devolução de um livro
def devolver_livro():
    print("\n--- DEVOLUÇÃO ---")
    titulo = input("Digite o título do livro: ").strip().lower()

    for livro in livros:
        if livro["titulo"].lower() == titulo:
            if livro["disponível"] == "Não":
                livro["disponível"] = "Sim"
                salvar_livros()
                print("Devolução realizada com sucesso!")
            else:
                print("Este livro já está disponível.")
            return

    print("Livro não encontrado.")


# Funcionalidade extra 1: contador do acervo
def contar_livros():
    total = len(livros)
    disponiveis = sum(
        1 for livro in livros
        if livro["disponível"] == "Sim"
    )
    emprestados = total - disponiveis

    print("\n--- CONTADOR DO ACERVO ---")
    print(f"Total de livros: {total}")
    print(f"Livros disponíveis: {disponiveis}")
    print(f"Livros emprestados: {emprestados}")


# Funcionalidade extra 2: ordenação alfabética
def ordenar_livros():
    if not livros:
        print("Nenhum livro cadastrado.")
        return

    ordenados = sorted(
        livros,
        key=lambda livro: livro["titulo"].lower()
    )

    print("\n--- LIVROS EM ORDEM ALFABÉTICA ---")

    for livro in ordenados:
        print(
            f"{livro['titulo']} - {livro['autor']} "
            f"({livro['ano']})"
        )


# Funcionalidade extra 3: remoção de livros
def remover_livro():
    print("\n--- REMOVER LIVRO ---")
    titulo = input("Digite o título do livro: ").strip().lower()

    for livro in livros:
        if livro["titulo"].lower() == titulo:
            if livro["disponível"] == "Não":
                print("Não é possível remover um livro emprestado.")
                return

            livros.remove(livro)
            salvar_livros()
            print("Livro removido com sucesso!")
            return

    print("Livro não encontrado.")


# Menu principal
def menu():
    carregar_livros()

    while True:
        print("\n===== SISTEMA DE BIBLIOTECA =====")
        print("1 - Cadastrar livro")
        print("2 - Listar livros")
        print("3 - Buscar livro")
        print("4 - Emprestar livro")
        print("5 - Devolver livro")
        print("6 - Contar livros")
        print("7 - Ordenar livros")
        print("8 - Remover livro")
        print("0 - Sair")

        opcao = input("Escolha uma opção: ").strip()

        if opcao == "1":
            cadastrar_livro()
        elif opcao == "2":
            listar_livros()
        elif opcao == "3":
            buscar_livro()
        elif opcao == "4":
            emprestar_livro()
        elif opcao == "5":
            devolver_livro()
        elif opcao == "6":
            contar_livros()
        elif opcao == "7":
            ordenar_livros()
        elif opcao == "8":
            remover_livro()
        elif opcao == "0":
            print("Sistema encerrado. Até logo!")
            break
        else:
            print("Opção inválida! Tente novamente.")


# Inicia o programa
if __name__ == "__main__":
    menu()
```
